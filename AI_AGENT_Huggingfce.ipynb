```python
from transformers import pipeline

# Initialize the sentiment analysis pipeline
# It automatically downloads the model and tokenizer on the first run
classifier = pipeline("sentiment-analysis")

# 1. Analyze a single sentence
result = classifier("I absolutely love using Hugging Face pipelines!")
print("Single Text Result:", result)

# 2. Analyze multiple sentences at once
sentences = [
    "The model configuration was a bit confusing at first.",
    "It works perfectly now and saves me so much time."
]
batch_results = classifier(sentences)

print("\nBatch Results:")
for text, res in zip(sentences, batch_results):
    print(f"Text: {text} -> Label: {res['label']}, Score: {res['score']:.4f}")
```

```python
import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

# 1. Load Model and Tokenizer
model_name = "facebook/bart-large-cnn"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# =====================================================================
# THE TEXT (Raw String Input)
# =====================================================================
text = """
The planetary system we call home is located in an outer spiral arm of the Milky Way galaxy.
Our solar system consists of our star, the Sun, and everything bound to it by gravity — the
planets Mercury, Venus, Earth, Mars, Jupiter, Saturn, Uranus, and Neptune; dwarf planets
such as Pluto; dozens of moons; and millions of asteroids, comets, and meteoroids. Beyond
our own solar system, there are more planets than stars in the night sky. So far, scientists
have discovered thousands of planetary systems orbiting other stars in the Milky Way,
with billions more waiting to be found.
"""

# =====================================================================
# THE INPUT (Tokenized Numerical Tensors)
# =====================================================================
# This converts your text words into mathematical IDs that the model understands
inputs = tokenizer(text, max_length=1024, truncation=True, return_tensors="pt")

print("--- MODEL INPUT TENSORS ---")
print("Input IDs (First 15 tokens):", inputs["input_ids"][0][:15])
print("Attention Mask (First 15 items):", inputs["attention_mask"][0][:15])
print(f"Total Input Shape: {inputs['input_ids'].shape}\n")

# =====================================================================
# GENERATION STEP
# =====================================================================
# The model processes the input tensors and returns numerical summary IDs
summary_ids = model.generate(
    inputs["input_ids"],
    num_beams=4,
    max_length=50,
    min_length=10,
    early_stopping=True
)

# =====================================================================
# THE SUMMARY (Decoded Human-Readable Output)
# =====================================================================
# Converts the numerical output IDs back into a readable string
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("--- GENERATED SUMMARY ---")
print(summary)
```

```python
import torch
from transformers import T5ForConditionalGeneration, T5Tokenizer

# 1. Load Model and Tokenizer
model_name = "t5-small"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)

# =====================================================================
# THE INPUT TEXT (With T5 Translation Prefix)
# =====================================================================
# T5 uses specific text prefixes to know which target language to use
raw_text = "The planetary system we call home is located in the Milky Way."
input_text = f"translate English to BENGALI: {raw_text}"

print("--- INPUT TEXT ---")
print(input_text, "\n")

# =====================================================================
# THE INPUT IDS (Tokenized Numerical Tensors)
# =====================================================================
inputs = tokenizer(input_text, return_tensors="pt")

print("--- MODEL INPUT TENSORS ---")
print("Input IDs Tensor:    ", inputs["input_ids"])
print("Attention Mask Tensor:", inputs["attention_mask"])
print(f"Input Shape:          {list(inputs['input_ids'].shape)}\n")

# =====================================================================
# GENERATION STEP
# =====================================================================
# The model processes the input IDs and generates translated token IDs
translated_ids = model.generate(
    **inputs,
    max_new_tokens=40,
    num_beams=4,
    early_stopping=True
)

print("--- OUTPUT IDS ---")
print("Generated Token IDs: ", translated_ids, "\n")

# =====================================================================
# GENERATED TRANSLATION (Decoded Output)
# =====================================================================
translation = tokenizer.decode(translated_ids[0], skip_special_tokens=True)

print("--- GENERATED TRANSLATION ---")
print(translation)
```

```python
import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

# 1. Load a model capable of handling Bengali translation
model_name = "facebook/nllb-200-distilled-600M"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# =====================================================================
# THE INPUT TEXT
# =====================================================================
input_text = "Tmy name is subho"

print("--- INPUT TEXT ---")
print(input_text, "\n")

# =====================================================================
# THE INPUT IDS
# =====================================================================
# NLLB requires setting the source language code (eng_Latn for English)
inputs = tokenizer(input_text, return_tensors="pt")

print("--- MODEL INPUT TENSORS ---")
print("Input IDs: ", inputs["input_ids"])
print(f"Shape:     {list(inputs['input_ids'].shape)}\n")

# =====================================================================
# GENERATION STEP
# =====================================================================
# We must explicitly tell the model to generate Bengali tokens (ben_Beng)
bengali_code_id = tokenizer.convert_tokens_to_ids("ben_Beng")

translated_ids = model.generate(
    **inputs,
    forced_bos_token_id=bengali_code_id,  # Forces Bengali output
    max_new_tokens=60,
    num_beams=4,
    early_stopping=True
)

# =====================================================================
# GENERATED TRANSLATION (Decoded Output)
# =====================================================================
translation = tokenizer.decode(translated_ids, skip_special_tokens=True)

print("--- GENERATED TRANSLATION (BENGALI) ---")
print(translation)
```

```python
import torch
from transformers import AutoModelForQuestionAnswering, AutoTokenizer

# 1. Load Model and Tokenizer
model_name = "google-bert/bert-large-uncased-whole-word-masking-finetuned-squad"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

# =====================================================================
# THE TEXT (Input Context Block)
# =====================================================================
text = """
The planetary system we call home is located in an outer spiral arm of the Milky Way galaxy.
Our solar system consists of our star, the Sun, and eight planets: Mercury, Venus, Earth, Mars,
Jupiter, Saturn, Uranus, and Neptune. It was formed 4.6 billion years ago from a massive
interstellar cloud.
"""

# =====================================================================
# THE QUESTIONS (Qs List)
# =====================================================================
questions = [
    "Where is our planetary system located?",
    "How many planets are in our solar system?",
    "When was the solar system formed?"
]

print("=== RUNNING EXTRACTIVE QA EVALUATION ===\n")

# =====================================================================
# GENERATION AND ANSWERS (As Extraction Loop)
# =====================================================================
for i, question in enumerate(questions, 1):
    # Prepare input text tensors (combining Question and Context together)
    inputs = tokenizer(question, text, return_tensors="pt")

    # Get the raw logit scores from the model
    with torch.no_grad():
        outputs = model(**inputs)

    # Find the positions of the highest-scoring start and end tokens
    answer_start = torch.argmax(outputs.start_logits)
    answer_end = torch.argmax(outputs.end_logits) + 1  # Add 1 to include the last token

    # Extract and decode the text span between those indices
    answer_tokens = inputs["input_ids"][0][answer_start:answer_end]
    answer = tokenizer.decode(answer_tokens, skip_special_tokens=True)

    # Clear out empty strings if the model can't find an answer span
    if not answer.strip():
        answer = "[No clear answer found in text]"

    print(f"--- BLOCK {i} ---")
    print(f"Q{i}: {question}")
    print(f"A{i}: {answer}\n")
```
